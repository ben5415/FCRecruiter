## 调用模型

````python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek
````

````python
load_dotenv(override=True)

# hello_message = os.getenv("HELLO_MESSAGE")
API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
````

````python
llm = ChatDeepSeek(
    model="deepseek-v4-flash", 
    api_key=API_KEY,  # 不指定则是DEEPSEEK_API_KEY
    base_url=BASE_URL  # 不指定则是DEEPSEEK_BASE_URL，也叫API_BASE
)
````

````python
response = llm.invoke("用一句话介绍自己")
````

````python
print(response)
````

````python
print(response.content)
````

### 直接 `ChatDeepSeek` vs `init_chat_model` 

````python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="deepseek-v4-flash",
    api_key=API_KEY,
    base_url=BASE_URL
)
````

``` python
model = init_chat_model(
    model="deepseek-v4-flash",
    api_key=API_KEY,
    base_url=BASE_URL
)
```
可以写成更全形式：
``` python
model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=API_KEY,
    base_url=BASE_URL
)
```
或者把 `model` 和 `model_provider` 二合一写成：
```python
model = init_chat_model(
    model="deepseek:deepseek-v4-flash",
    api_key=API_KEY,
    base_url=BASE_URL
)
```


````python
model_response = model.invoke("用一句话介绍一下 Football Manager 26 这款游戏")
````

````python
print(model_response.content)
````
